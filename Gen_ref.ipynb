```python
import pandas as pd, json
df = pd.read_csv('matatu_merged_clean.csv')
origins = df.groupby('travel_from')[['route','route_count','centroid_lat','centroid_lon','gis_missing']].first()
ref = {
    'origins': origins.to_dict('index'),
    'car_types': df.groupby('car_type')['max_capacity'].first().to_dict(),
    'payment_methods': sorted(df['payment_method'].unique()),
    'months': sorted(df['month'].unique()),
    'by_hour': df.groupby('hour')['is_overcrowded'].mean().to_dict(),
    'by_route': df.groupby('route')['is_overcrowded'].mean().to_dict(),
}
with open('app_reference.json', 'w') as f:
    json.dump(ref, f, indent=2, default=lambda o: o.item())
print('Saved app_reference.json')
```
